#Query the Endpoint via AI Gateway
This cell calls the **Databricks AI Gateway** using the OpenAI SDK's Responses API, routing requests to the `gpt-6-luna` model through the `dev.default.gpt-6-luna` provider service. Authentication uses the notebook's built-in session token via the notebook context API — no hardcoded credentials required.

For setup details, see [Create model provider services with Azure service credentials](https://learn.microsoft.com/en-us/azure/databricks/ai-gateway/create-model-provider-services#azure-service-credential).

In the example below, we deployed **gpt-6-luna** in Microsoft Foundry and registered it .  The cell below references this deployment and the model being registered in Unity Catalog as **dev.default.gpt-6-luna**.

```python
from openai import OpenAI

# Use notebook context token — works on all compute types including serverless
DATABRICKS_TOKEN = dbutils.notebook.entry_point.getDbutils().notebook().getContext().apiToken().get()

client = OpenAI(
  api_key=DATABRICKS_TOKEN,
  base_url="https://<workspace url>/ai-gateway/openai/v1",
  default_headers={"Databricks-Model-Provider-Service": "dev.default.gpt-6-luna"}
)

response = client.chat.completions.create(
    model="gpt-6-luna",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "What is the population of New York City?"}
    ],
)

print(response.choices[0].message.content)
```
